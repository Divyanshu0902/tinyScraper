```python
import random

z1 = random.random()
print(z1)

z2 = random.randint(1,100)
print(z2)

z3 = random.randrange(5,200,5)
print(z3)

z4 = random.uniform(2,7)
print(z4)

z5 = random.choice(['a','b','c','d','e'])
print(z5)

z6 = random.choices(['a','b','c','d','e'], k=2)
print(z6)

z7 = random.sample(['a','b','c','d','e'], k=2)
print(z7)

list1 = ['a','b','c','d','e']
random.shuffle(list1)
print(list1)

```
